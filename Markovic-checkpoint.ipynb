```python
import numpy as np
import pandas as pd
import math
import datetime as dt
import yfinance as yf
from pypfopt.efficient_frontier import EfficientFrontier
import matplotlib.pyplot as plt
from scipy.stats import norm
```

```python
portfolio = ['TSM', 'AAL', 'KWHIY', 'BABA', 'KO', 'PFE']
# TSMC, American Airlines, Kawasaki, Alibaba, Coca-Cola, Pfizer

# Restriccion maxima en portfolio:

if len(portfolio) > 10:
    raise Excepcion ('El número máximo de activos en el portfolio es de 10')

else:
    # Se cumple restricción:

    stock_prices = pd.DataFrame()
    stock_returns = pd.DataFrame()

    for x in portfolio:
        stock = yf.Ticker(x)
        close_price = stock.history(period='5y')['Close']

        # Se insertaran datos en stock_prices y stock_returns
        stock_prices = pd.concat([stock_prices, close_price], axis = 1)
        # Con pct_change() obtenemos los rendimientos es decir el cambio porcentual
        stock_returns = pd.concat([stock_returns, close_price.pct_change()], axis = 1)


# Asignamos nombre de acciones a las columnas de cada DataFrame

stock_prices.columns = portfolio
stock_returns.columns = portfolio
```

```python
stock_prices.head()
```

```python
stock_returns.dropna(inplace = True)
stock_returns.head()
```

```python
# Rendimientos esperados de activos
expected_stock_returns = []

# Riesgo individual de activos
individual_stock_risk = []

for x,y in stock_returns.items():
    # En cada itereción se obtiene el rendimiento esperado y riesgo individual de cada activo (media y desviación típica)
    expected_stock_returns.append(y.mean())
    individual_stock_risk.append(y.std())

# Tratamiento de datos para ser operados como vectores más adelante
expected_stock_returns = np.array(pd.DataFrame(expected_stock_returns))
individual_stock_risk = np.array(pd.DataFrame(individual_stock_risk))

# Ver datos obtenidos
print('Rendimientos esperados')
print(expected_stock_returns)
print('Riesgo individual')
print(individual_stock_risk)
```

```python
# Obtención de matriz de covarianzas

stock_returns_cov_matrix = np.array(stock_returns.cov())

# Ver resultados

print('Matriz de covarianzas de los rendimientos de los activos')
print(stock_returns_cov_matrix)
```

```python
# Creamos una cartera subóptima

def suboptimal_investment_portfolio(portfolio):
    suboptimal_portfolio = []

    # Iteramos y asignamos proporción igualitaria de inversión a cada activo del portfolio
    for x in portfolio:
        suboptimal_portfolio.append(1/len(portfolio))

    suboptimal_portfolio = np.array(suboptimal_portfolio)
    return suboptimal_portfolio

# Ejecutamos la función recién creada usando la cartera inicial como argumento
suboptimal_portfolio = suboptimal_investment_portfolio(portfolio)

# Tratamiento en numpy para futuras operacines
suboptimal_portfolio = np.expand_dims(suboptimal_portfolio, axis = 0)
# Ver los resultados
print('Cartera subóptima')
print(suboptimal_portfolio)
```

```python
# Rendimiento esperado de la cartera subóptima
expected_subopt_return = np.matmul(suboptimal_portfolio, expected_stock_returns)

# Varianza de la cartera 
subopt_portfolio_var = np.matmul(
    suboptimal_portfolio, 
    np.matmul(stock_returns_cov_matrix, suboptimal_portfolio.transpose())
)

# Riesgo de la cartera
subopt_portfolio_risk = np.sqrt(subopt_portfolio_var)

# Resultados
print('Cartera subóptima')
print(f'Rendimeinto esperado: {expected_subopt_return*100}')
print(f'Varianza: {subopt_portfolio_var*100}')
print(f'Volatilidad de la cartera: {subopt_portfolio_risk*100}') # Multiplica por 100 para que todos los resltados sean porcentajes
```

```python
# Optimización de Markovic
ef = EfficientFrontier(expected_stock_returns, stock_returns_cov_matrix, weight_bounds = (0, 1))
ratios = ef.min_volatility()
cleaned_ratios = pd.Series(ratios)
cleaned_ratios.index = portfolio

optimal_portfolio = np.expand_dims(cleaned_ratios, axis = 0)

# Rendimiento esperado de la cartera óptima
opt_portfolio_expected_return = np.matmul(optimal_portfolio, expected_stock_returns)

# Varianza de la cartera 
opt_portfolio_var = np.matmul(
    optimal_portfolio, 
    np.matmul(stock_returns_cov_matrix, optimal_portfolio.transpose())
)

# Riesgo de la cartera
opt_portfolio_risk = np.sqrt(opt_portfolio_var)

# Resultados
print('Cartera óptima de Markovic')
print(f'Rendimeinto esperado: {opt_portfolio_expected_return*100}')
print(f'Varianza: {opt_portfolio_var*100}')
print(f'Volatilidad de la cartera: {opt_portfolio_risk*100}')
print('\n\n Del 100% de tu capital, el modelo suguiere\
invertir las siguientes cantidades en cada activo:')
display(cleaned_ratios*100)
```

```python
# Optimización de Markovic
ef2 = EfficientFrontier(expected_stock_returns, stock_returns_cov_matrix, weight_bounds = (0, 1))
ratios2 = ef2.max_sharpe()
cleaned_ratios2 = pd.Series(ratios2)
cleaned_ratios2.index = portfolio

optimal_portfolio2 = np.expand_dims(cleaned_ratios2, axis = 0)

# Rendimiento esperado de la cartera óptima
opt_portfolio_expected_return2 = np.matmul(optimal_portfolio2, expected_stock_returns)

# Varianza de la cartera 
opt_portfolio_var2 = np.matmul(
    optimal_portfolio2, 
    np.matmul(stock_returns_cov_matrix, optimal_portfolio2.transpose())
)

# Riesgo de la cartera
opt_portfolio_risk2 = np.sqrt(opt_portfolio_var2)

# Resultados
print('Cartera óptima de Markovic')
print(f'Rendimeinto esperado: {opt_portfolio_expected_return2*100}')
print(f'Varianza: {opt_portfolio_var2*100}')
print(f'Volatilidad de la cartera: {opt_portfolio_risk2*100}')
print('\n\n Del 100% de tu capital, el modelo suguiere\
invertir las siguientes cantidades en cada activo:')
display(cleaned_ratios2*100)
```

```python
import matplotlib.pyplot as plt
from pypfopt import plotting

# 1. Configurar la figura
fig, ax = plt.subplots(figsize=(10, 6))

# 2. Trazar la Frontera Eficiente continua
ef_plot = EfficientFrontier(expected_stock_returns, stock_returns_cov_matrix, weight_bounds=(0, 1))
plotting.plot_efficient_frontier(ef_plot, ax=ax, show_assets=True)

# 3. Calcular métricas para ubicar los puntos clave
# Cartera Máximo Sharpe
ef_sharpe = EfficientFrontier(expected_stock_returns, stock_returns_cov_matrix, weight_bounds=(0, 1))
ef_sharpe.max_sharpe()
ret_sharpe, vol_sharpe, _ = ef_sharpe.portfolio_performance()

# Cartera Mínima Volatilidad
ef_minvol = EfficientFrontier(expected_stock_returns, stock_returns_cov_matrix, weight_bounds=(0, 1))
ef_minvol.min_volatility()
ret_minvol, vol_minvol, _ = ef_minvol.portfolio_performance()

# 4. Superponer los puntos de las carteras
ax.scatter(vol_sharpe, ret_sharpe, marker="*", s=250, c="gold", edgecolors="black", label="Máximo Sharpe")
ax.scatter(vol_minvol, ret_minvol, marker="o", s=120, c="blue", edgecolors="black", label="Mínima Volatilidad")
ax.scatter(subopt_portfolio_risk.item(), expected_subopt_return.item(), marker="X", s=120, c="red", label="Cartera 1/N (Subóptima)")

# 5. Ajustes estéticos
ax.set_title("Frontera Eficiente de Markowitz", fontsize=14)
ax.set_xlabel("Volatilidad (Riesgo)", fontsize=12)
ax.set_ylabel("Rendimiento Esperado", fontsize=12)
ax.legend()
ax.grid(True, linestyle="--", alpha=0.6)

plt.show()
```
